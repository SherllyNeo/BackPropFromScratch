# Facts used in this document

These are facts used throughout the examples, take them for granted if you wish or look them up if you are skeptical

* The derivative of a function with respect to a variable is how much that function changes when the variable changes, this is useful to find velocity which is a change in position or acceleration which is a change in velocity or whatever else

* The derivative of $y = f(x)$ can be written as $\frac{dy}{dx} = f'(x)$, but often only one side is used like so $\frac{dy}{dx}$, $f'(x)$ and you can also write the deriative as $y^{'}$, $\frac{d(f(x))}{dx}$ or $dy = f'(x) dx$
* The product rule let's us find the deriative of multiplied functions, $y=f(x)g(x)$, $y' = f g' + g f'$

* The derivative of a power such as $x^9$ is $9x^8$, that is always $y = x^n$, $\frac{dy}{dx} = n x^{n-1}$
* The deriative of $\sin{x}$ is $\cos{x}$
* The deriative of $\cos{x}$ is $-\sin{x}$
* We can flip deriatives, say we know $\frac{dx}{dy}$ and want to know $\frac{dy}{dx}$, we just flip it $\frac{dy}{dx} = \frac{1}{\frac{dx}{dy}}$. This can be useful in some chain rule questions
* $\frac{dy}{dx}$ is the total deriative, if $y$ is a function of several variables but we only want to see how it changes with $x$, we can use the partial deriative (which assumes all variables but $x$ are constant), we denote this $\frac{\partial y}{\partial x}$
* ReLU is linear for positive inputs, and sets anything negative to 0

# Introduction to the chain rule

Say we have a function $y = g(t)$ and we plug it into another function $t = f(x)$ to get


$y = f(g(x))$

How do we find $\frac{dy}{dx}$

The abuse of notation that is often used is $\frac{dy}{dx} = \frac{dy}{dt} \frac{dt}{dx}$

which equivalent to a statement we can prove later (in the video) $\frac{dy}{dx} = f'(x) g'(f(x))$

It looks like these cancel out, the same way fractions do 

$\frac{1}{3} = \frac{1}{4} \frac{4}{3}$

$\frac{1}{3} = \frac{1}{\not{4}} \frac{\not{4}}{3}$

$\frac{1}{3} = \frac{1}{3}$


The four here does not matter, you can use whatever number you want as it will cancel


This can be mimiced with the differentials

$\frac{dy}{dx} = \frac{dy}{dt} \frac{dt}{dx}$

$\frac{dy}{dx} = \frac{dy}{\not{dt}} \frac{\not{dt}}{dx}$

$\frac{dy}{dx} = \frac{dy}{dx}$

The trick with the chain rule is finding a suitable $t$ to differentiate 

This is the chain rule
$\frac{dy}{dx} = \frac{dy}{dt} \frac{dt}{dx}$

$y = (3x + 1)^3$

## Example 4

$\frac{dh}{dv} = \frac{1}{\pi 9m^2} $

$\frac{dh}{dt} = (\frac{1}{\pi 9m^2}) (0.5 \frac{m^3}{s})$

$\frac{dh}{dt} = \frac{0.5}{\pi 9}\frac{m}{s}$

And we can clean this up a little $0.5 = \frac{1}{2}$

$\frac{dh}{dt} = \frac{\frac{1}{2}}{\pi 9}\frac{m}{s}$

$\frac{dh}{dt} = \frac{1}{\pi 18}\frac{m}{s}$

## Example 6

We have a neural network that takes in two inputs $x_1$ and $x_2$

It has one hidden layer with output $o = w_1 x_1 + w_2 x_2 + b$

$o$ is wrapped in an activation function, we call this output $y = \sin(o)$, generally irl we use ReLU, but for simplicity we will use sine

The error $e$ is determined by the squared error function, the true output from the training data is $\hat{y}$

$e = (\hat{y} - y)^2 $

In order to update the weight $w_1$, we need to know how much the error $e$ changes when $w_1$ changes, that is $\frac{de}{dw_1}$

We can then update it, by choosing a new weight $w_1^{*} = w_1 - \alpha \frac{de}{dw_1}$, $\alpha$ is the learning rate, just a number we pick to ensure we scale the deriative so it does not jump around too much. 

we find $\frac{de}{dw_1}$ using the chain rule

$\frac{de}{dw_1} = \frac{de}{?} \frac{?}{dw_1}$

what do we put here? That's right

$\frac{de}{dw_1} = \frac{de}{dy} \frac{dy}{dw_1}$

The first term is easy

$\frac{de}{dy} = \frac{d(\hat{y} - y)^2}{dy} = -2(\hat{y} - y)$ (By the chain rule again, you should be able to do ones like this with ease after some practice)

For the second term $\frac{dy}{dw_1}$

To find this, we need the chain rule again

$\frac{dy}{dw_1} = \frac{dy}{do} \frac{do}{dw_1}$

$\frac{dy}{do} = \frac{d(\sin(o))}{do} = \cos(o)$

$\frac{do}{dw_1} = \frac{d( w_1 x_1 + w_2 x_2 + b)}{dw_1} = x_1 $

$\frac{dy}{dw_1} = \cos(o) x_1 $

so putting this together, we can calculate how much the output error depends on weight one $w_1$

$\frac{de}{dw_1} = (-2(\hat{y} - y))(\cos(o) x_1)$

and we know what $o$ is

$\frac{de}{dw_1} = (-2(\hat{y} - y))(\cos(w_1 x_1 + w_2 x_2 + b) x_1)$

it is important to note, these are all just numbers. We know our desired output $\hat{y}$, and the neural network output $y$ and all the weights, $w_1,w_2$ and the inputs $x_1,x_2$ and the bias $b$. 

So we can plug in all these numbers, and adjust $w_1$ accordingly, doing this for each weight and bias *is* gradient decent.

# Neural networks and backpropagation


The central idea of backprop is simple, we have a neural network, during the training step we give it some input $x$ and get some output $\hat{y}$, and we know we want $y$, we use some error function to tell us how much error there is between the neural network output and our desired value $e = E(y,\hat{y})$

So in order to improve our neural network, for each weight $w$ and bias $b$, we find how much the error changes with respect to them $\frac{de}{dw}$, $\frac{de}{db}$ and we use this to adjust those weights and biases

$w_{new} = w - \alpha \frac{de}{dw}$

$b_{new} = b - \alpha \frac{de}{db}$

where $\alpha$ is the learning rate, it controls how much we 'step' by. 

We can view the error function as the height in a rolling landscape of hills, where each co-ordinate is our particular choice of weights and biases, this adjust them in steps so that we walk down the hill to a valley, decreasing the error. This is not a metaphor, but mathematically what is going on.

Of course, we can get stuck in local minimums, there may be a deeper valley just over the hill but this method is likely to get stuck in the first valley it finds.

The way we find how the error function changes with respect to a particular weight or bias is with the **chain rule**!

## Real example of a neural network

**WARNING: Neural networks require messy index notation or obfuscated vector notation. No real away around it other than to wade through**

The notation used is two fold, there is the specific weights, biases and neuron outputs used for granular detail and then a vector notation that is more useful. 

The granular notation has conventions

$a^{(n)}_{j}$ this is the output of the $j$th neuron in the $n$th layer

$b^{(n)}_{j}$ this is the bias of the $j$th neuron in the $n$th layer

$w^{(n)}_{j,k}$ is the weight from the $k$th neuron in the $n-1$st layer to the $j$th neuron in the $n$th layer. The j and k may seem backwards, however it is useful for matrix stuff later.
$w^{(layer)}_{to,from}$

$z^{(n)}_{j}$ this is the weighted sum $j$th neuron in the $n$th layer, calculated from the incoming weights and activations/inputs as follows 

$z^{(n)}_{j} = a^{(n-1)}_{1} w^{(n)}_{j,1} + a^{(n-1)}_{2} w^{(n)}_{j,2} + \dots + a^{(n-1)}_{m} w^{(n)}_{j,m} + b^{(n)}_{j} $

We then use this to calculate the output

$a^{(n)}_{j} = f(z^{(n)}_{j})$

where $f$ is the activation function, this has to be differentiable for backpropagation to work.

The output of the neural network is $\hat{y}$ and we want it to be $y$, we then use an error function $e=E(y,\hat{y})$ to determine how much error the neural network has. Mean Squared Error is often used. Then see how much that error changes as we vary the parameters (weights and biases) to lower the error.

$a^{(n)}_{j} = f(z^{(n)}_{j})$

### Backpropagation

How does backprop work? Just the chain rule, say we want to update the weight

$w^{(1)}_{2,1}$, we need to find how the error function changes with respect to this weight

$\frac{\partial e}{\partial w^{(1)}_{2,1}}$

Then we can adjust the weight, once we have chosen a learning rate $\alpha$

$\text{new_w}^{(1)}_{2,1} = w^{(1)}_{2,1} - \alpha \frac{\partial e}{\partial w^{(1)}_{2,1}}$

So we need to find the deriative of the error, which is just a huge nested function

$e = (\hat{y}-y)^2$

$\frac{\partial e}{\partial w^{(1)}_{2,1}} = \frac{\partial e}{\partial \hat{y}} \frac{\partial \hat{y}}{\partial w^{(1)}_{2,1}}$

The first term  $\frac{\partial e}{\partial \hat{y}}$ is easy, $ \frac{\partial e}{\partial \hat{y}} =  \frac{\partial ( (\hat{y}-y)^2)}{\partial \hat{y}} = 2(\hat{y}-y)$

The second term... $\frac{\partial \hat{y}}{\partial w^{(1)}_{2,1}}$ a picture is helpful

![Three paths shown](backprop.jpg)


As you can see there are three paths to the weight we are interested in, which means we add up three partial derivatives

$\frac{\partial \hat{y}}{\partial w^{(1)}_{2,1}} = \frac{\partial \hat{y}}{\partial a^{(2)}_1} \frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_2} \frac{\partial a^{(2)}_2}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_3} \frac{\partial a^{(2)}_3}{\partial w^{(1)}_{2,1}}$

so we get

$\frac{\partial e}{\partial w^{(1)}_{2,1}} = \frac{\partial e}{\partial \hat{y}} \left( \frac{\partial \hat{y}}{\partial a^{(2)}_1} \frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_2} \frac{\partial a^{(2)}_2}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_3} \frac{\partial a^{(2)}_3}{\partial w^{(1)}_{2,1}} \right)$

Let's focus on the first term in the brackets

$ \frac{\partial \hat{y}}{a^{(2)}_1} \frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}}$

How do we compute this? Well the first bit ($\frac{\partial \hat{y}}{\partial a^{(2)}_1}$) is easier, recall that $\hat{y}$ is the activation of the output neuron, say it has activation function $\sigma(x)$ (In our example network, it is actually the identity function so $\sigma(x) = x$)

$\hat{y} = a^{(3)}_1 =  \sigma(z^{(3)}_1)$

$z^{(3)}_1 = b^{(3)}_1 + w^{(3)}_{1,1} a^{(2)}_1 + w^{(3)}_{1,2} a^{(2)}_2 + w^{(3)}_{1,3} a^{(2)}_3   $

so

$\frac{\partial \hat{y}}{\partial a^{(2)}_1}  = \frac{\partial \left( \sigma{\left( b^{(3)}_1 + w^{(3)}_{1,1} a^{(2)}_1 + w^{(3)}_{1,2} a^{(2)}_2 + w^{(3)}_{1,3} a^{(2)}_3  \right)}  \right)}{\partial a^{(2)}_1}$

Take some time convincing yourself of the next step, so we use the chain rule, we take the deriative of $\sigma$ which is $\sigma^{'}$ and multiply it by the deriative of $z^{(3)}_1$ with respect to ${a^{(2)}_1}$, recalling constants go to zero, the only term that actually matters is $w^{(3)}_{1,1} a^{(2)}_1$

$\frac{\partial \hat{y}}{\partial a^{(2)}_1}  = \sigma^{'}{ \left(z^{(3)}_1 \right)} w^{(3)}_{1,1} $

And we are done, we can compute the deriative of $\sigma^{'}$, we will know the weighted sum $z^{(3)}_1$ to plug into it and we will know the current weight value of $w^{(3)}_{1,1}$. We can just compute this number.

The second bit $\frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}}$ is more involved, we need the chain rule again


$\frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}} = \frac{\partial a^{(2)}_1}{\partial z^{(2)}_1} \frac{\partial z^{(2)}_1}{\partial w^{(1)}_{2,1}}$

$\frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}} = f'(z^{(2)}_1)  \frac{\partial z^{(2)}_1}{\partial w^{(1)}_{2,1}}$

recall $z^{(2)}_1 = b^{(2)}_1 + w^{(2)}_{1,1} a^{(1)}_1 +  w^{(2)}_{1,2} a^{(1)}_2 +  w^{(2)}_{1,3} a^{(1)}_3$

so

$\frac{\partial z^{(2)}_1}{\partial w^{(1)}_{2,1}} = \frac{\partial}{\partial w^{(1)}_{2,1}} \left( b^{(2)}_1 + w^{(2)}_{1,1} a^{(1)}_1 +  w^{(2)}_{1,2} a^{(1)}_2 +  w^{(2)}_{1,3} a^{(1)}_3 \right)$

most of these terms have no relation to the weight of interest $w^{(1)}_{2,1}$, so this simplifies nicely

$\frac{\partial z^{(2)}_1}{\partial w^{(1)}_{2,1}} = w^{(2)}_{1,2} \frac{\partial a^{(1)}_2}{\partial w^{(1)}_{2,1}}$

now recall 

$a^{(1)}_2 = f{\left( z^{(1)}_2 \right)} = f{\left( b^{(1)}_2 +  w^{(1)}_{2,1} a^{(0)}_1 \right)}$

There is only one term here because it is the connection to the input node, which products the inital activation.

convince yourself that

$\frac{\partial a^{(1)}_2}{\partial w^{(1)}_{2,1}} = f^{'}{\left( z^{(1)}_2  \right)} \frac{\partial z^{(1)}_2}{\partial w^{(1)}_{2,1}} $


$\frac{\partial a^{(1)}_2}{\partial w^{(1)}_{2,1}} = f^{'}{\left( z^{(1)}_2  \right)} \frac{\partial \left( b^{(1)}_2 +  w^{(1)}_{2,1} a^{(0)}_1  \right)}{\partial w^{(1)}_{2,1}} $


$\frac{\partial a^{(1)}_2}{\partial w^{(1)}_{2,1}} = f^{'}{\left( z^{(1)}_2  \right)}  a^{(0)}_1 $

so

$\frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}} = f'(z^{(2)}_1) w^{(2)}_{1,2} f^{'}{\left( z^{(1)}_2  \right)}  a^{(0)}_1 $


Therefore, the inital thing we were calculating

$\frac{\partial e}{\partial w^{(1)}_{2,1}}$

can be written as 

$\frac{\partial e}{\partial w^{(1)}_{2,1}} = 2(\hat{y}-y) \left( \frac{\partial \hat{y}}{\partial a^{(2)}_1} \frac{\partial a^{(2)}_1}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_2} \frac{\partial a^{(2)}_2}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_3} \frac{\partial a^{(2)}_3}{\partial w^{(1)}_{2,1}} \right)  $

which can be written as 

$\frac{\partial e}{\partial w^{(1)}_{2,1}} = 2(\hat{y}-y)\left( \left( \sigma^{'}{ \left(z^{(3)}_1 \right)} w^{3}_{1,1} f^{'}(z^{(2)}_1)  w^{(2)}_{1,2} f^{'}{\left( z^{(1)}_2  \right)} a^{(0)}_1 \right) + \frac{\partial \hat{y}}{\partial a^{(2)}_2} \frac{\partial a^{(2)}_2}{\partial w^{(1)}_{2,1}} + \frac{\partial \hat{y}}{\partial a^{(2)}_3} \frac{\partial a^{(2)}_3}{\partial w^{(1)}_{2,1}} \right)  $

We know $y$ and $\hat{y}$, all of the $z$, $w$ and $f^{'}(z)$ and such for each term, so we can calculate some of this.
As in, we can just plug in the numbers for these variables.

Okay so all of this was just for one of the three terms we want, but the others work in the same way every time we move backwards through a neuron, we multiply by the derivative of its activation function and the relevant connection weight. Eventually, we reach the weight we're interested in, where the derivative of its weighted sum contributes the input to that weight.


![Chain rule applied to a neural network error function](chainrule.jpg)

Once we fully calculate this, we can update our weight of interest $w^{(1)}_{2,1}$ via the rule

$\text{new_w}^{(1)}_{2,1} = w^{(1)}_{2,1} - \alpha \frac{\partial e}{\partial w^{(1)}_{2,1}}$


# Vector Notation

The notation above is terrible, we will use vector notation mostly

$\mathbf{w}^{(n)}$ is a matrix of all the weights connecting the $n-1$st layer to the $n$th layer

$\mathbf{b}^{(n)}$ is a vector of all the biases for the $n$th layer.

$\mathbf{z}^{(n)}$ is a vector of all the weighted summations for the $n$th layer.

$\mathbf{a}^{(n)}$ is a vector of all the activations for the $n$th layer.

## Backprop using vector notation

The backprop algorithm can be condensed into 4 equations

1. **Output layer error using the chain rule**

$\mathbf{\delta}^{(L)} = \frac{\partial e}{\partial \mathbf{a}^{(L)}} \odot (f^{(L)})'(\mathbf{z}^{(L)})$

2. **Backpropagate the error using more chain rule**

$\mathbf{\delta}^{(n)} = ((\mathbf{w}^{(n+1)})^T\mathbf{\delta}^{(n+1)}) \odot f'(\mathbf{z}^{(n)})$

3. **Calculate the weight and bias gradients with even more chain rule**

$\frac{\partial e}{\partial \mathbf{w}^{(n)}} = \mathbf{\delta}^{(n)}(\mathbf{a}^{(n-1)})^T$

$\frac{\partial e}{\partial \mathbf{b}^{(n)}} = \mathbf{\delta}^{(n)}$

4. **Update the weights and biases - this is gradient decent**

$\mathbf{w}^{(n)}_{\text{new}} = \mathbf{w}^{(n)} - \alpha \frac{\partial e}{\partial \mathbf{w}^{(n)}}$

$\mathbf{b}^{(n)}_{\text{new}} = \mathbf{b}^{(n)} - \alpha \frac{\partial e}{\partial \mathbf{b}^{(n)}}$


We can write backpropagation using vector notation, which allows us to calculate the gradients for every weight and bias in a layer at once.

We use the chain rule, just as before, but instead of calculating the derivative of each weight individually, we do them a layer at a time.

## 1. The generic case

Our forward pass can be written as 

$\mathbf{z}^{(n)} = \mathbf{w}^{(n)}\mathbf{a}^{(n-1)} + \mathbf{b}^{(n)}$

$\mathbf{a}^{(n)} = f(\mathbf{z}^{(n)})$

and our error function is

$e = (\hat{y}-y)^2$

We want to find the derivative of the error with respect to the weights and biases.

###  Calculate the output error

We start with the derivative of the error with respect to the predicted output.

$\frac{\partial e}{\partial \hat{y}} = 2(\hat{y}-y)$

For the output layer, we define the error signal as

$\mathbf{\delta}^{(3)} = \frac{\partial e}{\partial \mathbf{z}^{(3)}}$

Using the chain rule,

$\mathbf{\delta}^{(3)} = \frac{\partial e}{\partial \hat{y}} \frac{\partial \hat{y}}{\partial \mathbf{z}^{(3)}}$

Since the output activation is the identity function, its derivative is $1$.

$\implies \mathbf{\delta}^{(3)} = 2(\hat{y}-y)$

More generally, if the output activation is $\sigma$, then

$\mathbf{\delta}^{(3)} = 2(\hat{y}-y)\sigma'(\mathbf{z}^{(3)})$

### Calculate the gradients for the output weights and bias

Recall that

$\mathbf{z}^{(3)} = \mathbf{w}^{(3)}\mathbf{a}^{(2)} + \mathbf{b}^{(3)}$

The gradient with respect to the output weights is

$\frac{\partial e}{\partial \mathbf{w}^{(3)}} = \mathbf{\delta}^{(3)}(\mathbf{a}^{(2)})^T$

The gradient with respect to the output bias is

$\frac{\partial e}{\partial \mathbf{b}^{(3)}} = \mathbf{\delta}^{(3)}$

We can now update the output weights and bias using the learning rate $\alpha$.

$\mathbf{w}^{(3)}_{\text{new}} = \mathbf{w}^{(3)} - \alpha \frac{\partial e}{\partial \mathbf{w}^{(3)}}$

$\mathbf{b}^{(3)}_{\text{new}} = \mathbf{b}^{(3)} - \alpha \frac{\partial e}{\partial \mathbf{b}^{(3)}}$

### Propagate the error backwards

We now calculate the error signal for the second hidden layer.

The error signal for layer $2$ is

$\mathbf{\delta}^{(2)} = \frac{\partial e}{\partial \mathbf{z}^{(2)}}$

Using the chain rule,

$\mathbf{\delta}^{(2)} = (\mathbf{w}^{(3)})^T\mathbf{\delta}^{(3)} \odot f'(\mathbf{z}^{(2)})$

Here, $\odot$ represents element-wise multiplication.

We multiply the error signal from the next layer by the transpose of the weights connecting the layers, then multiply element-wise by the derivative of the activation function.

We can now calculate the gradients for the second layer weights and biases.

$\frac{\partial e}{\partial \mathbf{w}^{(2)}} = \mathbf{\delta}^{(2)}(\mathbf{a}^{(1)})^T$

$\frac{\partial e}{\partial \mathbf{b}^{(2)}} = \mathbf{\delta}^{(2)}$

And update them:

$\mathbf{w}^{(2)}_{\text{new}} = \mathbf{w}^{(2)} - \alpha \frac{\partial e}{\partial \mathbf{w}^{(2)}}$

$\mathbf{b}^{(2)}_{\text{new}} = \mathbf{b}^{(2)} - \alpha \frac{\partial e}{\partial \mathbf{b}^{(2)}}$

### Propagate the error to the first hidden layer

We repeat the same process to calculate the error signal for the first hidden layer.

$\mathbf{\delta}^{(1)} = \frac{\partial e}{\partial \mathbf{z}^{(1)}}$

$\mathbf{\delta}^{(1)} = (\mathbf{w}^{(2)})^T\mathbf{\delta}^{(2)} \odot f'(\mathbf{z}^{(1)})$

We can then calculate the gradients for the first layer weights and biases.

$\frac{\partial e}{\partial \mathbf{w}^{(1)}} = \mathbf{\delta}^{(1)}(\mathbf{a}^{(0)})^T$

$\frac{\partial e}{\partial \mathbf{b}^{(1)}} = \mathbf{\delta}^{(1)}$

And update them

$\mathbf{w}^{(1)}_{\text{new}} = \mathbf{w}^{(1)} - \alpha \frac{\partial e}{\partial \mathbf{w}^{(1)}}$

$\mathbf{b}^{(1)}_{\text{new}} = \mathbf{b}^{(1)} - \alpha \frac{\partial e}{\partial \mathbf{b}^{(1)}}$

We have now calculated the gradients for every weight and bias in the network in one fell swoop. Much nicer than looping over every weight and bias

## 2. Apply this to our network (Used chatGPT here, it isn't super important to plug the numbers in but might help understanding)

Recall our network has input $x=2$ and is modelling $y=3x^2+5$.

Our forward pass gave us

$\mathbf{a}^{(0)} = \begin{bmatrix} 2 \end{bmatrix}$

$\mathbf{a}^{(1)} = \begin{bmatrix} 1.5 \\ 0 \\ 0.92 \end{bmatrix}$

$\mathbf{a}^{(2)} = \begin{bmatrix} 0 \\ 1.173 \\ 0.5632 \end{bmatrix}$

$\hat{y} = -0.049588$

The true output is

$y = 3(2)^2+5 = 17$

Our error is

$e = (\hat{y}-y)^2$

$\implies e = (-0.049588-17)^2$

$\implies e = 291.688458\ldots$

### Step 1: Output error signal

Since our output activation is the identity function,

$\delta^{(3)} = 2(\hat{y}-y)$

$\implies \delta^{(3)} = 2(-0.049588-17)$

$\implies \delta^{(3)} = -34.099176$

### Step 2: Output weight gradients

The gradient of the output weights is

$\frac{\partial e}{\partial \mathbf{w}^{(3)}} = \delta^{(3)}(\mathbf{a}^{(2)})^T$

$\implies \frac{\partial e}{\partial \mathbf{w}^{(3)}} = (-34.099176)\begin{bmatrix} 0 & 1.173 & 0.5632 \end{bmatrix}$

$\implies \frac{\partial e}{\partial \mathbf{w}^{(3)}} = \begin{bmatrix} 0 & -39.998334648 & -19.2054560832 \end{bmatrix}$

$\frac{\partial e}{\partial b^{(3)}} = \delta^{(3)}$

$\implies \frac{\partial e}{\partial b^{(3)}} = -34.099176$

For ReLU, the derivative is

$f'(z) = \begin{cases} 1 & z>0 \\ 0 & z<0 \end{cases}$

Our second hidden layer weighted sums were

$\mathbf{z}^{(2)} = \begin{bmatrix} -0.5172 \\ 1.173 \\ 0.5632 \end{bmatrix}$

Therefore,

$f'(\mathbf{z}^{(2)}) = \begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix}$

The error signal is

$\mathbf{\delta}^{(2)} = (\mathbf{w}^{(3)})^T\delta^{(3)} \odot f'(\mathbf{z}^{(2)})$

$\implies \mathbf{\delta}^{(2)} = \begin{bmatrix} -0.6 \\ -0.1 \\ -0.59 \end{bmatrix}(-34.099176) \odot \begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix}$

$\implies \mathbf{\delta}^{(2)} = \begin{bmatrix} 20.4595056 \\ 3.4099176 \\ 20.11851384 \end{bmatrix} \odot \begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix}$

$\implies \mathbf{\delta}^{(2)} = \begin{bmatrix} 0 \\ 3.4099176 \\ 20.11851384 \end{bmatrix}$

### Step 4: Second hidden layer weight gradients

$\frac{\partial e}{\partial \mathbf{w}^{(2)}} = \mathbf{\delta}^{(2)}(\mathbf{a}^{(1)})^T$

$\implies \frac{\partial e}{\partial \mathbf{w}^{(2)}} = \begin{bmatrix} 0 \\ 3.4099176 \\ 20.11851384 \end{bmatrix}\begin{bmatrix} 1.5 & 0 & 0.92 \end{bmatrix}$

$\implies \frac{\partial e}{\partial \mathbf{w}^{(2)}} = \begin{bmatrix} 0 & 0 & 0 \\ 5.1148764 & 0 & 3.136124192 \\ 30.17777076 & 0 & 18.5090527328 \end{bmatrix}$

The bias gradients are

$\frac{\partial e}{\partial \mathbf{b}^{(2)}} = \mathbf{\delta}^{(2)}$

$\implies \frac{\partial e}{\partial \mathbf{b}^{(2)}} = \begin{bmatrix} 0 \\ 3.4099176 \\ 20.11851384 \end{bmatrix}$

Therefore,

The bias gradients are

$\implies \frac{\partial e}{\partial \mathbf{b}^{(1)}} = \begin{bmatrix} -1.061454648 \\ 0 \\ 5.5954776744 \end{bmatrix}$

## 3. Update the weights and biases

Once we choose a learning rate $\alpha$, we update all the weights and biases using gradient descent.

For example, the first layer weights become

$\mathbf{w}^{(1)}_{\text{new}} = \mathbf{w}^{(1)} - \alpha \frac{\partial e}{\partial \mathbf{w}^{(1)}}$

$\implies \mathbf{w}^{(1)}_{\text{new}} = \begin{bmatrix} 0.8 \\ -0.03 \\ 0.01 \end{bmatrix} - \alpha \begin{bmatrix} -2.122909296 \\ 0 \\ 11.1909553488 \end{bmatrix}$

The same process applies to every other weight matrix and bias vector.

```python
# Generate data
layer_size = 64
sample_size = 1000

x_train = np.random.uniform(-10, 10, (sample_size, 1))
noise = np.random.normal(0, 1, (sample_size, 1))
y_train = 3 * x_train**2 + 5 + noise

# Arrange the data so each column is one training sample
x = x_train.T
y = y_train.T


# Initialise weights and biases

W_1 = np.random.uniform(-1, 1, (1, layer_size))
b_1 = np.random.uniform(-1, 1, (layer_size, 1))

W_2 = np.random.uniform(-1, 1, (layer_size, layer_size))
b_2 = np.random.uniform(-1, 1, (layer_size, 1))

W_3 = np.random.uniform(-1, 1, (layer_size, 1))
b_3 = np.random.uniform(-1, 1, (1, 1))

alpha = 0.00001 # Tiny learning rate or it blows up
epochs = 50000


def ReLU(z):
    return np.maximum(0, z)


def ReLU_derivative(z):
    return (z > 0).astype(float)


for epoch in range(epochs):

    # 1. FORWARD PASS
    # z^(n) = W^(n).T @ a^(n-1) + b^(n)
    # a^(n) = f(z^(n))

    # Input layer: a^(0)
    a_0 = x

    # First hidden layer
    z_1 = W_1.T @ a_0 + b_1
    a_1 = ReLU(z_1)

    # Second hidden layer
    z_2 = W_2.T @ a_1 + b_2
    a_2 = ReLU(z_2)

    # Output layer (identity activation)
    z_3 = W_3.T @ a_2 + b_3
    y_hat = z_3

    # We have our output, time to calcualte the error

    # e = (y_hat - y)^2
    error = y_hat - y
    mse = np.mean(error**2)


    # 2. OUTPUT ERROR SIGNAL
    # delta^(3) = 2(y_hat - y) * sigma'(z^(3))

    delta_3 = 2 * error / sample_size


    # 3. BACKPROPAGATE THE ERROR
    # delta^(n) = (W^(n+1) @ delta^(n+1)) * f'(z^(n))

    # Error signal for the second hidden layer
    delta_2 = (W_3 @ delta_3) * ReLU_derivative(z_2)

    # Error signal for the first hidden layer
    delta_1 = (W_2 @ delta_2) * ReLU_derivative(z_1)


    
    # 4. CALCULATE THE GRADIENTS
    # dE/dW^(n) = a^(n-1) @ delta^(n).T
    # dE/db^(n) = sum(delta^(n), axis=1)


    # Output layer gradients
    dEwithrespecttodW_3 = a_2 @ delta_3.T
    dEwithrespecttodb_3 = np.sum(delta_3, axis=1, keepdims=True)

    # Second hidden layer gradients
    dEwithrespecttodW_2 = a_1 @ delta_2.T
    dEwithrespecttodb_2 = np.sum(delta_2, axis=1, keepdims=True)

    # First hidden layer gradients
    dEwithrespecttodW_1 = a_0 @ delta_1.T
    dEwithrespecttodb_1 = np.sum(delta_1, axis=1, keepdims=True)


    # 5. UPDATE THE PARAMETERS
    W_1 -= alpha * dEwithrespecttodW_1
    b_1 -= alpha * dEwithrespecttodb_1

    W_2 -= alpha * dEwithrespecttodW_2
    b_2 -= alpha * dEwithrespecttodb_2

    W_3 -= alpha * dEwithrespecttodW_3
    b_3 -= alpha * dEwithrespecttodb_3

    if epoch % 5000 == 0:
        print(f"Epoch {epoch} | Training MSE: {mse}")


# Test
x_test = np.linspace(-10, 10, 100).reshape(-1, 1)
y_test = 3 * x_test**2 + 5

a_0 = x_test.T

# Forward pass using the trained parameters
z_1 = W_1.T @ a_0 + b_1
a_1 = ReLU(z_1)

z_2 = W_2.T @ a_1 + b_2
a_2 = ReLU(z_2)

z_3 = W_3.T @ a_2 + b_3
y_pred = z_3.T
```

```python
# Eval
test_mse = np.mean((y_pred - y_test)**2)
print(f"Test MSE: {test_mse}")


sample_predictions = []

for i in range(0, len(x_test), 10):
    actual = y_test[i, 0]
    predicted = y_pred[i, 0]

    sample_predictions.append({
        "Input": x_test[i, 0],
        "Actual": actual,
        "Predicted": predicted,
        "Abs Error": abs(actual - predicted)
    })

df_predictions = pd.DataFrame(sample_predictions)

df_predictions
```

```python
def predict():
    x_input = float(input("Enter a value for x: \n(Works best on the range it was trained on [-10,10])\n"))

    # Forward pass
    a_0 = np.array([[x_input]])

    z_1 = W_1.T @ a_0 + b_1
    a_1 = ReLU(z_1)

    z_2 = W_2.T @ a_1 + b_2
    a_2 = ReLU(z_2)

    z_3 = W_3.T @ a_2 + b_3
    y_hat = z_3[0, 0]

    # Calculate the real answer
    y = 3 * x_input**2 + 5
    error = abs(y_hat - y)

    print(f"\nInput:          {x_input}")
    print(f"Predicted:      {y_hat}")
    print(f"Real answer:    {y}")
    print(f"Absolute error: {error}")


predict()
```
